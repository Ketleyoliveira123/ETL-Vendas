```python
import pandas as  pd
```

```python
df_cliente  = pd.read_csv(r"C:\Users\Ketle\OneDrive\Documentos\projeto_py\SalesFlow-ETL\data\processed\clientes.csv")
df_produto  = pd.read_csv(r"C:\Users\Ketle\OneDrive\Documentos\projeto_py\SalesFlow-ETL\data\processed\produtos.csv")
df_venda    = pd.read_csv(r"C:\Users\Ketle\OneDrive\Documentos\projeto_py\SalesFlow-ETL\data\processed\vendas.csv")
```

```python
# Juntar vendas com clientes
df_final = df_venda.merge(df_cliente, on="id_cliente", how="left")

# Juntar o resultado com produtos
df_final = df_final.merge(df_produto, on="id_produto", how="left")

```

```python
# Conferir o resultado
display(df_final)
print("Dimensões:", df_final.shape)
print("Valores ausentes por coluna:")
print(df_final.isnull().sum())
```

```python
print(df_final[df_final.isnull().any(axis=1)])

```

```python
# corrija a quantidade e confira novamente os valores ausentes
df_final['quantidade'] = df_final['quantidade'].fillna(0)
print("Valores ausentes por coluna após preenchimento:")
print(df_final.isnull().sum())
```
