```python
import pandas as pd
import sqlite3

df = pd.read_csv("../data/processed/vendas_integradas.csv")

with sqlite3.connect("../data/salesflow.db") as conn:
    df.to_sql("vendas", conn, if_exists="replace", index=False)

print("Dados carregados com sucesso!")
```

```python
with sqlite3.connect("../data/salesflow.db") as conn:
    print(pd.read_sql_query("SELECT COUNT(*) AS total FROM vendas LIMIT 5", conn)) # esse comando é apenas para conferir se os dados foram carregados corretamente. Ele retorna o total de registros na tabela "vendas".
```
