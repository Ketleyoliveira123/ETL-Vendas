```python
import pandas as pd

df_cliente=pd.read_csv(r"C:\Users\Ketle\OneDrive\Documentos\projeto_py\SalesFlow-ETL\data\raw\clientes.csv")
df_produto = pd.read_csv(r"C:\Users\Ketle\OneDrive\Documentos\projeto_py\SalesFlow-ETL\data\raw\produtos.csv")
df_venda = pd.read_csv(r"C:\Users\Ketle\OneDrive\Documentos\projeto_py\SalesFlow-ETL\data\raw\vendas.csv")  
```

```python
# Converter quantidade para número
df_venda['quantidade'] = pd.to_numeric(
    df_venda['quantidade'],
    errors='coerce'
)

df_venda = df_venda.drop_duplicates(keep='first')

# Limpar nomes dos clientes
df_cliente['nome'] = df_cliente['nome'].str.strip().str.title() # Padronizar nomes dos clientes

# Padronizar cidades
df_cliente['cidade'] = df_cliente['cidade'].str.strip().str.title()
```

```python
from pathlib import Path

# Caminho da pasta processed
pasta_saida = Path(r"C:\Users\Ketle\OneDrive\Documentos\projeto_py\SalesFlow-ETL\data\raw\processed")
pasta_saida.mkdir(parents=True, exist_ok=True)

# Salvar os DataFrames tratados
df_venda.to_csv(pasta_saida / "vendas.csv", index=False, encoding="utf-8")
df_cliente.to_csv(pasta_saida / "clientes.csv", index=False, encoding="utf-8")
df_produto.to_csv(pasta_saida / "produtos.csv", index=False, encoding="utf-8")

```
